````python
import math

class Planet:
    """
    A class representing a planet orbiting a star.

    Attributes:
        mass (float): The mass of the planet.
        orbital_radius (float): The distance of the planet from its star (meters).
        star_mass (float): The mass of the star the planet orbits (kg).
    """
    def __init__(self, mass, orbital_radius, star_mass):
        """
        Initialize a Planet instance with mass, orbital radius, and the mass of its star.
        """
        self.mass = mass
        self.orbital_radius = orbital_radius
        self.star_mass = star_mass
        
    def get_orbital_period(self):
        """
        Calculate and return the orbital period of the planet using Kepler's Third Law.
        """
        G = 6.67430e-11  # Gravitational constant in m^3/kg/s^2
        T = 2 * math.pi * math.sqrt((self.orbital_radius ** 3) / (G * self.star_mass))
        return T / (60 * 60 * 24)  # Convert seconds to days
````

## Modeling a system of 2 planets around a star

If we have a system of 2 planets orbiting a star, these planets interact with each other gravitationally. This force is proportional to the planets' masses (bigger mass -> stronger force/interaction) and inversely proportional to the distance between the planets (bigger distance -> weaker force/interaction). If this force is substantial, it can move the planets from their initial orbits, even make them collide in extreme cases! To make sense of how this happens, we can start from a simplified model of 2 planets orbiting the same star. In this part you will create and run that model, plot how the radius from the host star changes for each planet.

````python
def calculate_forces(planet1, planet2):
    """
    Calculate the gravitational force that planet2 exerts on planet1.

    This function computes the radial force component based on the gravitational interaction 
    between two planets. 

    Parameters:
        planet1 (object): The first planet (experiencing the force).
        planet2 (object): The second planet (exerting the force).

    Returns:
        float: The radial force acting on planet1 due to planet2.
    """
    force_x = planet2.mass * planet1.mass * (planet2.x - planet1.x) / ((planet2.x - planet1.x)**2 + (planet2.y - planet1.y)**2)**(1.5)
    force_y = planet2.mass * planet1.mass * (planet2.y - planet1.y) / ((planet2.x - planet1.x)**2 + (planet2.y - planet1.y)**2)**(1.5)
    force_radial = force_x * planet1.x / planet1.radius + force_y * planet1.y / planet1.radius
    return force_radial


def take_step(planet, force, delta_time):
    """
    Update the planet's position, velocity, and radius over a time step.

    This function calculates how much the planet moves in the given time step 
    due to the applied force. 

    Parameters:
        planet (object): The planet being updated.
        force (float): The radial force acting on the planet.
        delta_time (float): The time step for the simulation.

    Returns:
        None: The function modifies the planet's properties directly.
    """
    acceleration = force / planet.mass
    radius_change = 0.5 * acceleration * delta_time**2 + planet.velocity_radial * delta_time
    velocity_change = acceleration * delta_time
    planet.radius = planet.radius + radius_change
    planet.x = planet.radius * np.cos(2 * np.pi * delta_time / planet.orbital_period)
    planet.y = planet.radius * np.sin(2 * np.pi * delta_time / planet.orbital_period)
    planet.velocity_radial = planet.velocity_radial + velocity_change
    
    if planet.radius < 0.001:
        planet.radius = 0
        planet.status = "Terminated"
    
    planet.orbital_period = 2 * np.pi * np.sqrt(planet.radius**3 / planet.star_mass)
````

````python
# your code here
import numpy as np
import math

class Planet:
    """
    Represents a planet orbiting a star.

    Attributes:
        mass (float): Mass of the planet.
        radius (float): Initial orbital radius of the planet.
        star_mass (float): Mass of the central star.
        orbital_period (float): Orbital period computed using Kepler's third law.
        velocity_radial (float): Initial radial velocity of the planet, set to 0.0.
        x (float): Initial x-coordinate of the planet.
        y (float): Initial y-coordinate of the planet.
        status (str): Status of the planet ("Active" or "Terminated").
    """

    def __init__(self, mass, radius, star_mass):
        """
        Initializes a Planet object with given mass, radius, and star mass.
        """
        self.mass = mass
        self.radius = radius
        self.star_mass = star_mass
        self.orbital_period = 2 * math.pi * math.sqrt(radius**3 / star_mass)  # Kepler's Third Law
        self.velocity_radial = 0.0
        self.x = radius
        self.y = 0.0
        self.status = "Active"

    def check_termination(self):
        """
        Checks if the planet's radius has fallen below sqrt(0.02),
        if so, updates status to "Terminated".
        """
        if self.radius < np.sqrt(0.02):
            self.status = "Terminated"


class System:
    """
    Represents a planetary system with a central star and two orbiting planets.

    Attributes:
        star_mass (float): Mass of the central star.
        planet1 (Planet): First planet in the system.
        planet2 (Planet): Second planet in the system.
        planet1_pos (list): Tracks the position of planet1 over time.
        planet2_pos (list): Tracks the position of planet2 over time.
    """

    def __init__(self, star_mass, planet1, planet2):
        """
        Initializes the System with a central star and two planets.
        """
        self.star_mass = star_mass
        self.planet1 = planet1
        self.planet2 = planet2
        self.planet1_pos = [[0.0, planet1.radius]]
        self.planet2_pos = [[0.0, planet2.radius]]

    def evolve(self, time):
        """
        Simulates the evolution of the system over the given time.

        Parameters:
            time (float): The total time duration for the simulation.

        Updates the planets' positions and tracks their radii over time.
        """
        t = 0.0
        delta_time = self.planet1.orbital_period / 5  # Small time step for accuracy

        while t < time:
            total_force_planet1 = 0.0
            total_force_planet2 = 0.0

            if (self.planet1.status == "Active" and self.planet2.status == "Active" and 
                (self.planet1.x - self.planet2.x)**2 + (self.planet1.y - self.planet2.y)**2 >= 0.02):

                # Compute gravitational forces between planets
                total_force_planet1 = calculate_forces(self.planet1, self.planet2)
                total_force_planet2 = calculate_forces(self.planet2, self.planet1)

                # Update positions of both planets
                take_step(self.planet1, total_force_planet1, delta_time)
                take_step(self.planet2, total_force_planet2, delta_time)

            # Increment time step
            t += delta_time

            # Record the positions of both planets
            self.planet1_pos.append([t, self.planet1.radius])
            self.planet2_pos.append([t, self.planet2.radius])

            # Check if any planet has terminated
            self.planet1.check_termination()
            self.planet2.check_termination()

            if self.planet1.status == "Terminated" or self.planet2.status == "Terminated":
                break  # Stop simulation if any planet crashes

````

### Modeling a system of 1 big planet and N small planets around a star

Now let's extend our model for a system of 1 big planet (by big we mean Jupiter size planet) and a bunch of small planets around it, which will be an approximation for the disk.

Here we will provide the generalized code that runs for a set number of planets.


````python

class System:
    def __init__(self, star_mass, planet_list):
        # the primaty planet we care about is the first planet in the list
        self.star_mass = star_mass
        
        self.planet_list = planet_list
        
        self.number_of_planets = len(planet_list)
        
        self.plot_positions = [[0.0, planet_list[0].radius]]
    
    def evolve(self, time):
        t = 0.0
        delta_time = self.planet_list[0].orbital_period/5
        while t < time:
            # delta_t = self.planet_list[0].orbital_period/5 ## this is the step that chooses the resolution of our simulation
            total_force_list = np.zeros(self.number_of_planets)
            for i, planet in enumerate(self.planet_list):
                if planet.status == "Active":
                    total_force_radial = 0.0
                    for other_planet in self.planet_list:
                        if other_planet.status == "Active":
                            if other_planet != planet and (other_planet.x - planet.x)**2 + (other_planet.y - planet.y)**2 >= 0.02:
                                total_force_radial += calculate_forces(planet, other_planet)
                    total_force_list[i] = total_force_radial

            for i, planet in enumerate(self.planet_list):
                if planet.status == "Active":
                    take_step(planet, total_force_list[i], delta_time)

            
            for planet in self.planet_list:
                if planet.radius < 0.02:
                    planet.status == "Terminated"
            if self.planet_list[0].status == "Terminated":
                break
            t += delta_time
            self.plot_positions.append([t, self.planet_list[0].radius])
            
````

````python
#your code here
import numpy as np
import math
import matplotlib.pyplot as plt

class Planet:
    """
    Represents a planet orbiting a star.
    """
    def __init__(self, mass, radius, star_mass):
        self.mass = mass
        self.radius = radius
        self.star_mass = star_mass
        self.orbital_period = 2 * math.pi * math.sqrt(radius**3 / star_mass)  # Kepler's Third Law
        self.velocity_radial = 0.0
        self.x = radius
        self.y = 0.0
        self.status = "Active"

    def check_termination(self):
        """
        Checks if the planet's radius has fallen below sqrt(0.02),
        if so, updates status to "Terminated".
        """
        if self.radius < np.sqrt(0.02):
            self.status = "Terminated"

class System:
    """
    Represents a planetary system with a central star and multiple orbiting planets.
    """
    def __init__(self, star_mass, planet_list):
        self.star_mass = star_mass
        self.planet_list = planet_list
        self.number_of_planets = len(planet_list)
        self.plot_positions = [[0.0, planet_list[0].radius]]

    def calculate_forces(self, planet1, planet2):
        """
        Calculate the gravitational force that planet2 exerts on planet1.
        """
        dx = planet2.x - planet1.x
        dy = planet2.y - planet1.y
        distance_squared = dx**2 + dy**2
        distance = math.sqrt(distance_squared)

        if distance < 0.001:  # Avoid division by very small numbers
            return 0

        force = (planet1.mass * planet2.mass) / distance_squared
        force_x = force * dx / distance
        force_y = force * dy / distance

        force_radial = force_x * planet1.x / planet1.radius + force_y * planet1.y / planet1.radius
        return force_radial

    def take_step(self, planet, force, delta_time):
        """
        Updates the planet’s position, velocity, and radius over a time step.
        """
        acceleration = force / planet.mass
        radius_change = 0.5 * acceleration * delta_time**2 + planet.velocity_radial * delta_time
        velocity_change = acceleration * delta_time
        planet.radius = max(planet.radius + radius_change, 0.001)  # Prevent negative radius
        planet.x = planet.radius * np.cos(2 * np.pi * delta_time / planet.orbital_period)
        planet.y = planet.radius * np.sin(2 * np.pi * delta_time / planet.orbital_period)
        planet.velocity_radial += velocity_change

        if planet.radius < np.sqrt(0.02):
            planet.status = "Terminated"

        planet.orbital_period = 2 * np.pi * np.sqrt(planet.radius**3 / planet.star_mass)

    def evolve(self, time):
        """
        Simulates the evolution of the system over the given time.
        """
        t = 0.0
        delta_time = self.planet_list[0].orbital_period / 5

        while t < time:
            total_force_list = np.zeros(self.number_of_planets)

            for i, planet in enumerate(self.planet_list):
                if planet.status == "Active":
                    total_force_radial = 0.0
                    for other_planet in self.planet_list:
                        if other_planet.status == "Active" and other_planet != planet:
                            if (other_planet.x - planet.x)**2 + (other_planet.y - planet.y)**2 >= 0.02:
                                total_force_radial += self.calculate_forces(planet, other_planet)
                    total_force_list[i] = total_force_radial

            for i, planet in enumerate(self.planet_list):
                if planet.status == "Active":
                    self.take_step(planet, total_force_list[i], delta_time)

            for planet in self.planet_list:
                if planet.radius < 0.02:
                    planet.status = "Terminated"

            if self.planet_list[0].status == "Terminated":
                break

            t += delta_time
            self.plot_positions.append([t, self.planet_list[0].radius])

primary_planet = Planet(mass=150, radius=1, star_mass=400)

num_small_planets = 50
small_planets = [
    Planet(mass=0.1, radius=r, star_mass=400) 
    for r in np.linspace(0.01, 0.7, num_small_planets)
]

planet_list = [primary_planet] + small_planets

system = System(star_mass=400, planet_list=planet_list)
system.evolve(time=10)

time_values = [pos[0] for pos in system.plot_positions]
primary_radii = [pos[1] for pos in system.plot_positions]

plt.figure(figsize=(8, 5))
plt.scatter(time_values, primary_radii, s=5, label="Primary Planet (Mass 150)", color="blue")

plt.xlabel("Time")
plt.ylabel("Orbital Radius")
plt.title("Orbital Radius vs. Time for Primary Planet in a 51-Planet System")
plt.legend()

plt.show()
````
